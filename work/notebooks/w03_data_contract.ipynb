## 2. Fields: feature / label / context / excluded

### Features

I will use measurable performance signals as candidate features:

- `gsc_avg_position` — average search position.

These features are measured performance signals that can support decision-making about content performance.

### Label

The label will represent whether a content page is declining over a defined future outcome window. It will be derived from a measured outcome rather than from a manually chosen feature threshold.

### Context

- `client_hash_id` — identifies the anonymized client.
- `content_hash_id` — identifies the anonymized content page.

These fields describe the observation or its data availability rather than directly serving as predictive features.

### Excluded

I will exclude fields that are not necessary for the first feature frame or that may create leakage.

In particular, I will exclude future-outcome information from the feature set. A field derived from the future decline outcome must not be used as a feature because it would reveal the answer to the model.

I will also avoid using `client_hash_id` and `content_hash_id` as predictive features because they are identifiers rather than behavioral measurements.

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

## 4. Data limits

This contract has several important limitations.

### 1. Unbalanced panel

The warehouse is an unbalanced panel, so different clients have different history depths. Therefore, the absence of a measurement does not necessarily mean that the underlying activity was zero.

### 2. Data-source availability

GSC and GA4 measurements are only available when the corresponding data source is available for the client and observation. In the March 2026 slice, the availability fields show that only a subset of rows has GSC or GA4 data available.

### 3. Missing feature values

`gsc_avg_position`, `ga4_pageviews`, and `ga4_sessions` contain missing values. These missing values should not automatically be converted to zero because missingness can represent unavailable data rather than zero activity.

### 4. Leakage risk

A future outcome or a feature derived from the future outcome must not be available at the decision moment. For example, a column created directly from a future decline label would allow a model to see the answer it is supposed to predict.

For this reason, future-derived outcome information will be excluded from the feature set.

### 5. Scope of this contract

The contract describes the available warehouse measurements and their limitations. It does not establish that the selected features cause content performance changes. The measurements are intended for decision-support and prediction experiments.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Final feature contract check

feature_contract = pd.DataFrame({
    "field": [
        "gsc_impressions",
        "gsc_clicks",
        "gsc_avg_position",
        "ga4_pageviews",
        "ga4_sessions"
    ],
    "role": ["feature"] * 5,
    "knowable_at_decision": [True] * 5,
    "missing_values": [
        march_df["gsc_impressions"].isna().sum(),
        march_df["gsc_clicks"].isna().sum(),
        march_df["gsc_avg_position"].isna().sum(),
        march_df["ga4_pageviews"].isna().sum(),
        march_df["ga4_sessions"].isna().sum()
    ]
})

display(feature_contract)
```

## 5. Future outcome label and leakage check

The March 2026 observation is treated as the decision-time data.

The future outcome window is 2026-04-02 through 2026-06-30. I will aggregate query-level impressions to the client-content level and compare the last-30-day impressions with the previous-30-day impressions.

The label `is_declining_label` is 1 when future last-30-day impressions are lower than future previous-30-day impressions, and 0 otherwise.

Because this label is calculated from future observations, it is available only as an outcome for evaluation. It must not be included among the March decision-time features.

A label-derived feature would create leakage because it would expose information from the outcome directly to the model.

```python
# Create a future content-level outcome label

future_outcome = (
    query_df
    .groupby(["client_hash_id", "content_hash_id"], as_index=False)
    .agg(
        future_impressions_last30=("impressions_last30", "sum"),
        future_impressions_prev30=("impressions_prev30", "sum")
    )
)

future_outcome["is_declining_label"] = (
    future_outcome["future_impressions_last30"]
    < future_outcome["future_impressions_prev30"]
).astype(int)

print("Future outcome rows:", len(future_outcome))

print("\nLabel counts:")
print(future_outcome["is_declining_label"].value_counts())

print("\nLabel rate:")
print(future_outcome["is_declining_label"].mean())
```

## 6. Deliberate label leakage experiment

To demonstrate leakage, I will temporarily create a feature directly from the future label.

This feature is intentionally invalid because it contains the answer the model is supposed to predict.

The purpose of this experiment is to show why label-derived information must be excluded from the decision-time feature set.

```python
# Deliberate leakage demonstration

leakage_demo = future_outcome[
    ["client_hash_id", "content_hash_id", "is_declining_label"]
].copy()

# INTENTIONALLY WRONG:
# This feature directly copies the future label.
leakage_demo["leaked_feature"] = leakage_demo["is_declining_label"]

print("Rows:", len(leakage_demo))

print("\nLabel vs leaked feature:")
print(
    leakage_demo[
        ["is_declining_label", "leaked_feature"]
    ].head(10)
)

print(
    "\nExact match between label and leaked feature:",
    (
        leakage_demo["is_declining_label"]
        == leakage_demo["leaked_feature"]
    ).mean()
)
```

## 7. Final honest feature contract

The leakage experiment showed that a feature derived directly from the future label reproduces the target exactly.

That feature is therefore removed from the final feature set.

The final features are limited to measurements available at the March 2026 decision moment:

- `gsc_impressions`
- `gsc_clicks`
- `gsc_avg_position`
- `ga4_pageviews`
- `ga4_sessions`

The future `is_declining_label` is retained only as the evaluation outcome and is not included as a model feature.

This separation prevents future information from entering the decision-time feature set.

```python
# Remove the intentionally leaked feature

honest_features = [
    "gsc_impressions",
    "gsc_clicks",
    "gsc_avg_position",
    "ga4_pageviews",
    "ga4_sessions"
]

final_contract = pd.DataFrame({
    "field": honest_features,
    "role": ["feature"] * len(honest_features),
    "decision_time_data": [True] * len(honest_features),
    "future_derived": [False] * len(honest_features)
})

print("Final honest feature set:")
display(final_contract)

print("\nExcluded leaked field:")
print("leaked_feature")

print("\nOutcome kept separately:")
print("is_declining_label")
```
